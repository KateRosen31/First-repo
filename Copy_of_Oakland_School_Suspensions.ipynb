```python
import requests
import time

key = #insert your API key here

def get_address(name, key):
    if name != str:
        name = str(name)
    base = "https://maps.googleapis.com/maps/api/geocode/json?address="
    url = base + name.replace(" ", "+") + "&key=" + key
    r = requests.get(url)

    try:
        return r.json()['results'][0]['formatted_address']
    except IndexError or KeyError:
        pass

address = []

for index in range(len(sdf.filename)):
    address.append((get_address(sdf.search_term[index],key)))
    time.sleep(0.5)

ad = pd.DataFrame(address, columns=['Address'])

print(ad)
```

```python
result = pd.concat([sdf, ad], axis=1, sort=False)
result['csv'] = '.csv'
result['filename.csv'] = result['filename'] + result['csv']
result.head()
```

```python
import plotly.graph_objects as go
import pandas as pd

ousd_data = [['2014-15', 4.05], ['2015-16', 4.07], ['2016-17', 3.99], ['2017-18', 4.29], ['2018-19', 4.37]]
black_ousd_data = [['2014-15', 8.31], ['2015-16', 8.44], ['2016-17', 8.02], ['2017-18', 9.23], ['2018-19', 8.74]]
public_ousd_data = [['2014-15', 3.8], ['2015-16', 3.94], ['2016-17', 3.82], ['2017-18', 4.12], ['2018-19', 3.95]]
charter_ousd_data = [['2014-15', 5.29], ['2015-16', 4.72], ['2016-17', 4.66], ['2017-18', 4.93], ['2018-19', 5.75]]

df1 = pd.DataFrame(ousd_data, columns = ['School Year', 'Suspension Rate'])
df2 = pd.DataFrame(black_ousd_data, columns = ['School Year', 'Suspension Rate'])
df3 = pd.DataFrame(public_ousd_data, columns=['School Year', 'Suspension Rate'])
df4 = pd.DataFrame(charter_ousd_data, columns=['School Year', 'Suspension Rate'])

fig = go.Figure()

fig.add_trace(go.Scatter(x=df3['School Year'], y=df3['Suspension Rate'], mode='lines+markers', name='Public Suspension Rate'))
fig.add_trace(go.Scatter(x=df4['School Year'], y=df4['Suspension Rate'], mode='lines+markers', name='Charter Suspension Rate'))

fig.update_xaxes(type='category')
fig.update_xaxes(tickangle = 90, title_text = 'School Year', title_standoff = 25)
fig.update_yaxes (title_text = 'Suspension Rate', title_standoff = 25)
fig.update_layout(template='seaborn')

fig.show()
fig.write_html('/Users/arijitsen/Desktop/Oakland Suspensions/public_vs_charter_scatter.html')
```
